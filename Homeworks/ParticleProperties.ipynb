```python
import numpy as np # import numpy
import astropy.units as u # import astropy units
from astropy import constants as const # import astropy constants
from ReadFile import Read
 # this is so we can read the file and take the data for our calculations
def ParticleInfo(filename, particle_type, particle_number):
    """ This function will compute the magnitude of the distance in kpc, magnitude of the velocity in km/s and mass in unit of solar mass.It will give the properties of any given particle type that the reader decides.
    Inputs: filename - the name of the file so it can be open and read for the Read function to be used
        particle type is the type of particle (Type 1 = Dark Matter, Type 2 = Disk Stars, Type 3 = Bulge Stars)
        particle number is the number of the particle that is being asked to be used
    Outputs: distance (kpc)- magnitude of the distance rounded 3 places,velocity(km/s)- magnitude of the velocity rounded 3 places,mass(solar mass)- the particle's mass
    """
    #read the data from the imported file
    time, number,data = Read(filename)
    # this is used to go where the particle type is located in the data easier (1-3)
    types = data['type'] == particle_type
    i = np.where(types)[0] # making sure it goes to the location for the info
    # this is now used for the particles number to be located exactly with the other code function
    pnumber= data[i[particle_number]]
    #this is to calcuate the distance magnitude |d|= sqrt(postionx^2+positony^2+postionz^2)
    dis = np.sqrt((pnumber['x']*u.kpc)**2+(pnumber['y']*u.kpc)**2+(pnumber['z']*u.kpc)**2)
    #this is to calculate the velocity magnitude |v|= sqrt(velocityx^2+velocityy^2+velocityz^2)
    vel= np.sqrt((pnumber['vx']*u.km/u.s)**2+(pnumber['vy']*u.km/u.s)**2+(pnumber['vz']*u.km/u.s)**2)
    distance = np.around(dis,3) #now we are just rounding 
    velocity = np.around(vel,3)# rounding again
    # mass is now being calculated which is just the particles number mass in the data times 1x10 solar mass
    mass= pnumber['m']*1e10        
    return distance, velocity, mass# final results 
```

```python
# using the code to find the 100 disk particle (2,101)
distance,velocity,mass=ParticleInfo('MW_000.txt',2,99) #using the function and since its asking for the 100th we subtract one more since it starts at 0
lightd=distance.to(u.lyr) # converting to light years 
lightyear=np.around(lightd,3) # rounding 
print('3D Distance: {} kpc and {} lightyears 3D Velocity: {} km/s Mass: {} Mo'.format(distance,lightyear,velocity,mass))
```
